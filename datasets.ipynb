```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'C:\Users\rahma khawaja\Downloads\pakistani_ecommerce_dataset.csv')
```

```python
df.head()
```

```python
most_popular_category = df['product_category'].mode()[0]
print("Q1 - Most popular category:", most_popular_category)
```

```python
avg_order_value = np.nanmean(df['order_value'])
print("Q2 - Average order value:", round(avg_order_value, 2))
```

```python
top_city = df['city'].value_counts().idxmax()
print("Q3 - City with highest orders:", top_city)
```

```python
top_payment = df['payment_method'].mode()[0]
print("Q4 - Most used payment method:", top_payment)
```

```python
# Q5

city_avg = df.groupby('city')['order_value'].mean()

city_avg.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Order Value by City')
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
avg_study = np.nanmean(df['study_hours'])
print("Q1 - Average study hours:", round(avg_study, 2))
```

```python
sleep_by_gender = df.groupby('gender')['sleep_hours'].mean()
print("\nQ2 - Average sleep hours by gender:")
print(sleep_by_gender)
print("Who sleeps more:", sleep_by_gender.idxmax())
```

```python
correlation = df['screen_time'].corr(df['sleep_hours'])
print("\nQ3 - Correlation between screen time and sleep:", round(correlation, 3))
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['daily_transport_cost'].dropna(), bins=10, color='teal', edgecolor='black')
plt.title('Distribution of Daily Transport Cost')
plt.xlabel('Daily Transport Cost')
plt.ylabel('Number of Students / People')
plt.grid(axis='y', alpha=0.5)
plt.tight_layout()
plt.show()
```
